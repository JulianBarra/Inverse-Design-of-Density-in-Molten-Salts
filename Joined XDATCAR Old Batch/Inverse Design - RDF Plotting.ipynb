```python
import pandas as pd
import matplotlib.pyplot as plt

# read text file into pandas DataFrame
one_str  = lambda x: '/Users/jebo2/OneDrive/Documents/Jupyter/Thesis Work/Joined XDATCAR Old Batch/'+x+'/data_one'
dec_str  = lambda x: '/Users/jebo2/OneDrive/Documents/Jupyter/Thesis Work/Joined XDATCAR Old Batch/'+x+'/data_dec'
r_str = 'Pair separation distance'

#List of different folders for the simulations (REMEMBER TO PUT LDMNS5)
list_l = ['HDDEF', 'HDMNS5', 'HDPLS15', 'HDPLS30', 'LDDEF', 'LDMNS15', 'LDPLS15', 'LDPLS30', 'MDDEF', 'MDMNS5', 'MDPLS15']

#df = pd.read_csv(dirstring, sep=" ", skiprows=[0], index_col=False)
```

```python
for l in list_l:
    one = pd.read_csv(one_str(l), sep=" ", skiprows=[0], index_col=False)
    dec = pd.read_csv(dec_str(l), sep=" ", skiprows=[0], index_col=False)
    plt.figure()
    for ll in list(dec.keys()[1:]):
        plt.plot(dec[r_str], dec[ll], label = ll)
    plt.plot(one[r_str], one['g(r)'], label = 'g(r)')
    plt.legend()
    plt.show()
```
